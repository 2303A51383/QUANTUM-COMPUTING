```python
from qiskit import QuantumCircuit
from qiskit_aer import AerSimulator
from qiskit.quantum_info import Statevector
from qiskit.visualization import plot_histogram, plot_bloch_multivector
import matplotlib.pyplot as plt
import numpy as np

simulator = AerSimulator()

print("Qiskit is ready!")
```

```python
qc = QuantumCircuit(1, 1)

qc.id(0)
qc.measure(0, 0)

print(qc)

result = simulator.run(qc, shots=1024).result()
counts = result.get_counts()

print("Measurement Results:", counts)
plot_histogram(counts)
plt.show()
```

```python
qc = QuantumCircuit(1)
qc.id(0)

state = Statevector.from_instruction(qc)

print("Statevector:", state)
plot_bloch_multivector(state)
plt.show()
```

```python
import ipywidgets as widgets
from IPython.display import display, clear_output
import matplotlib.pyplot as plt
import numpy as np
from qiskit import QuantumCircuit
from qiskit.quantum_info import Statevector
from qiskit.visualization import plot_bloch_multivector, plot_histogram
from qiskit_aer import AerSimulator
circuit_names = [
    "Identity (I)",
    "Pauli-X",
    "Pauli-Y",
    "Pauli-Z",
    "Hadamard (H)",
    "Phase-S",
    "Phase-T",
    "RX",
    "RY",
    "RZ",
    "Measurement",
    "CNOT (CX)",
    "CZ",
    "SWAP",
    "Toffoli (CCX)",
    "Bell State",
    "GHZ State"
]
dropdown = widgets.Dropdown(
    options=circuit_names,
    description="Circuit:",
    style={'description_width': 'initial'},
    layout=widgets.Layout(width='350px')
)
button = widgets.Button(
    description="Run Circuit",
    button_style='primary',
    layout=widgets.Layout(width='200px')
)
output = widgets.Output()
def run_circuit(b):
    with output:
        clear_output()
        name = dropdown.value
        if name == "Identity (I)":
            qc = QuantumCircuit(1)
            qc.id(0)
        elif name == "Pauli-X":
            qc = QuantumCircuit(1)
            qc.x(0)
        elif name == "Pauli-Y":
            qc = QuantumCircuit(1)
            qc.y(0)
        elif name == "Pauli-Z":
            qc = QuantumCircuit(1)
            qc.z(0)
        elif name == "Hadamard (H)":
            qc = QuantumCircuit(1)
            qc.h(0)
        elif name == "Phase-S":
            qc = QuantumCircuit(1)
            qc.s(0)
        elif name == "Phase-T":
            qc = QuantumCircuit(1)
            qc.t(0)
        elif name == "RX":
            qc = QuantumCircuit(1)
            theta = np.pi / 2

            qc.rx(theta, 0)

        elif name == "RY":

            qc = QuantumCircuit(1)

            theta = np.pi / 2

            qc.ry(theta, 0)

        elif name == "RZ":

            qc = QuantumCircuit(1)

            theta = np.pi / 2

            qc.rz(theta, 0)


        elif name == "Measurement":

            qc = QuantumCircuit(1)

            qc.h(0)



        elif name == "CNOT (CX)":

            qc = QuantumCircuit(2)

            qc.x(0)
            qc.cx(0, 1)


        elif name == "CZ":

            qc = QuantumCircuit(2)

            qc.h(0)
            qc.h(1)
            qc.cz(0, 1)


        elif name == "SWAP":

            qc = QuantumCircuit(2)

            qc.x(0)
            qc.swap(0, 1)

        elif name == "Toffoli (CCX)":

            qc = QuantumCircuit(3)

            qc.x(0)
            qc.x(1)
            qc.ccx(0, 1, 2)

        elif name == "Bell State":

            qc = QuantumCircuit(2)

            qc.h(0)
            qc.cx(0, 1)



        elif name == "GHZ State":

            qc = QuantumCircuit(3)

            qc.h(0)
            qc.cx(0, 1)
            qc.cx(1, 2)


        print("=" * 60)

        print("Selected Circuit:")
        print(name)

        print("=" * 60)


        print("\nCircuit Diagram:")

        print(qc.draw(output='text'))



        state = Statevector.from_instruction(qc)

        print("\nStatevector:")

        print(state)



        print("\nState Visualization:")


        if state.num_qubits == 1:

            display(plot_bloch_multivector(state))

        else:

            print("Bloch sphere is shown only for single-qubit circuits.")




        measured_qc = qc.copy()

        measured_qc.measure_all()


        simulator = AerSimulator()


        result = simulator.run(
            measured_qc,
            shots=1024
        ).result()


        counts = result.get_counts()


        print("\nMeasurement Results:")

        print(counts)



        print("\nMeasurement Histogram:")


        fig = plot_histogram(
            counts,
            title=f"{name} - Measurement Results"
        )

        display(fig)


        print("\n" + "=" * 60)

        print("Circuit execution completed!")

        print("=" * 60)




button.on_click(run_circuit)


display(dropdown)

display(button)

display(output)
```
